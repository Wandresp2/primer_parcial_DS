```python
import mysql.connector

conn = mysql.connector.connect(
    host="mysql",
    port=3306,
    user="root",
    password="root",
    database="db_prueba",
)

cursor = conn.cursor()
cursor.execute("SELECT VERSION();")
print(cursor.fetchone())
cursor.close()
conn.close()
```
