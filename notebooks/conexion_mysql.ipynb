# Prueba de conexion Jupyter a MySQL

El host debe ser el nombre del servicio en Docker Compose (`mysql`), no `localhost`.

Credenciales segun `.env`: usuario `root`, password `root`, base `db_prueba`.

```python
%pip install mysql-connector-python
```

```python
import mysql.connector

conn = mysql.connector.connect(
    host="mysql",
    port=3306,
    user="root",
    password="root",
    database="db_prueba",
)

cursor = conn.cursor()
cursor.execute("SELECT VERSION();")
print("Conexion exitosa. Version MySQL:", cursor.fetchone()[0])

cursor.execute("SELECT DATABASE();")
print("Base de datos actual:", cursor.fetchone()[0])

cursor.close()
conn.close()
print("Conexion cerrada correctamente.")
```
